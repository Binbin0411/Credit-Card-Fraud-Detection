```python
import joblib
import pandas as pd
```

```python
model = joblib.load("models/fraud_model.pkl")
scaler = joblib.load("models/scaler.pkl")

print("Model loaded successfully.")
print("Scaler loaded successfully.")
```

```python
df = pd.read_csv("creditcard.csv")

df.head()
```

```python
sample = df.drop("Class", axis=1).iloc[[100]]
actual = df["Class"].iloc[100]

print(sample)
```

```python
sample_scaled = scaler.transform(sample)
```

```python
prediction = model.predict(sample_scaled)

print("Prediction:", prediction[0])
print("Actual:", actual)
```

```python
probability = model.predict_proba(sample_scaled)

print(probability)
```

```python
if prediction[0] == 1:
    print("🚨 Fraud Transaction")
else:
    print("✅ Normal Transaction")

print("--------------------------------")

print("Probability")

print("Normal :", probability[0][0])

print("Fraud  :", probability[0][1])

print("--------------------------------")

print("Actual :", actual)
```

```python
samples = df.drop("Class", axis=1).iloc[0:10]

actual = df["Class"].iloc[0:10]

samples_scaled = scaler.transform(samples)

prediction = model.predict(samples_scaled)

result = pd.DataFrame({

    "Actual": actual.values,

    "Prediction": prediction

})

result
```
